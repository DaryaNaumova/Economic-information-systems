```python
import yfinance as yf
import pandas as pd

# Определение тикера
ticker = 'AAPL'

# Извлечение данных
data = yf.download(ticker, start='2016-01-01', end='2023-01-01')

# Вывод первых строк данных
print(data.head())
```

```python
import matplotlib.pyplot as plt

# Предполагаем, что данные уже загружены в DataFrame 'data'
# Проверяем, что столбец 'Adj Close' существует в DataFrame
if 'Adj Close' in data.columns:
    # Построение графика скорректированной цены закрытия
    data['Adj Close'].plot(figsize=(10, 6))
    plt.title('Скорректированная Цена Закрытия')
    plt.xlabel('Дата')
    plt.ylabel('Цена')
    plt.grid()  # Добавляем сетку для лучшей читаемости графика
    plt.show()
else:
    print("Столбец 'Adj Close' не найден в данных.")
```
