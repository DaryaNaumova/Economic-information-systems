```python
import pandas as pd

# Создание данных
data = {
    'name': ['Саша', 'Маша', 'Соня', 'Миша', 'Вова'],
    'score': [5, 4, 3, 4, 5]
}

# Создание DataFrame
df = pd.DataFrame(data)

# Сохранение DataFrame в CSV файл
file_path = 'data.csv'
df.to_csv(file_path, index=False)

print(f"Файл '{file_path}' успешно создан.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Загрузка данных из CSV файла
def load_data(file_path):
    try:
        data = pd.read_csv(file_path)
        print("Данные успешно загружены.")
        return data
    except FileNotFoundError:
        print("Файл не найден. Проверьте путь к файлу.")
        return None

# Обработка данных: вычисление среднего значения
def process_data(data):
    mean_score = data['score'].mean()
    print(f"Средний балл: {mean_score:.2f}")
    return mean_score

# Визуализация данных
def visualize_data(data):
    plt.figure(figsize=(10, 6))
    plt.bar(data['name'], data['score'], color='skyblue')
    plt.xlabel('Имя')
    plt.ylabel('Баллы')
    plt.title('Баллы учеников')
    plt.axhline(y=data['score'].mean(), color='r', linestyle='--', label='Средний балл')
    plt.legend()
    plt.show()

# Основная функция
def main():
    file_path = 'data.csv'  # Путь к файлу с данными
    data = load_data(file_path)

    if data is not None:
        process_data(data)
        visualize_data(data)

if __name__ == "__main__":
    main()
```

## PyDrive

Ниже вы найдете примеры использования PyDrive для аутентификации и загрузки/скачивания файлов. Больше примеров можно найти в <a href="https://pythonhosted.org/PyDrive/">документации PyDrive</a>.

```python
# Импортируем класс GoogleAuth из библиотеки PyDrive для аутентификации в Google Drive.
from pydrive.auth import GoogleAuth

# Импортируем класс GoogleDrive из библиотеки PyDrive для работы с Google Drive API.
from pydrive.drive import GoogleDrive

# Импортируем модуль auth из google.colab для выполнения аутентификации в Google.
from google.colab import auth

# Импортируем класс GoogleCredentials из oauth2client.client для работы с учетными данными Google API.
from oauth2client.client import GoogleCredentials
```

Выполните аутентификацию и создайте клиент PyDrive.


```python
auth.authenticate_user() # Аутентифицируем пользователя через интерфейс, предоставляемый библиотекой
gauth = GoogleAuth() # Инициализируем экземпляр GoogleDrive, передавая ему аутентифицированные учетные данные
gauth.credentials = GoogleCredentials.get_application_default() # Инициализируем экземпляр GoogleDrive, передавая ему аутентифицированные учетные данные
drive = GoogleDrive(gauth) # Создаём новый экземпляр GoogleAuth (генерируем новую сессию аутентификации)
```

## Drive REST API

Чтобы использовать Drive API, сначала нужно выполнить аутентификацию и создать клиент API.


```python
from google.colab import auth # Импортируем модуль auth из google.colab для аутентификации пользователя
auth.authenticate_user() # Аутентифицируем пользователя, включая все необходимые разрешения для доступа к Google Drive
from googleapiclient.discovery import build # Импортируем модуль build из googleapiclient.discovery для взаимодействия с API Google
drive_service = build('drive', 'v3') # Создаём клиентский объект для доступа к сервису Google Drive версии 3 (v3)
```

С этим клиентом мы можем использовать любые функции из <a href="https://developers.google.com/drive/v3/reference/">документации по Google Drive API</a>. Примеры представлены ниже.


### Создание файла на Диске на основе данных из Python

Прежде всего, создайте локальный файл для загрузки.

```python
with open('/tmp/to_upload.txt', 'w') as f: # Открываем файл с именем 'to_upload.txt' для записи ('w') в временной директории '/tmp'
  f.write('my sample file') # Записываем строку 'my sample file' в открытый файл

print('/tmp/to_upload.txt contains:') # Выводим сообщение о содержимом файла '/tmp/to_upload.txt'
!cat /tmp/to_upload.txt # Используем команду shell для отображения содержимого файла 'to_upload.txt'
```

Загрузите его, используя метод <a href="https://developers.google.com/drive/v3/reference/files/create"><code>files.create</code></a>. Более подробная информация о загрузке файлов доступна в <a href="https://developers.google.com/drive/v3/web/manage-uploads">документации для разработчиков</a>.

```python
from googleapiclient.http import MediaFileUpload # Импортируем класс MediaFileUpload из библиотеки googleapiclient.http для загрузки файлов

file_metadata = {
  'name': 'Sample file',
  'mimeType': 'text/plain' # Указываем MIME-тип для корректной обработки файла
} # Создаём словарь с метаданными файла, включая имя файла и его MIME-тип
media = MediaFileUpload('/tmp/to_upload.txt',
                        mimetype='text/plain',
                        resumable=True) # Создаём объект MediaFileUpload с указанием пути к файлу и его MIME-типом
created = drive_service.files().create(body=file_metadata, # Метаданные файла
                                       media_body=media, # Загружаемое содержимое файла
                                       fields='id').execute()  # Запрашиваем только идентификатор созданного файла
print('File ID: {}'.format(created.get('id')))
```

После выполнения кода в ячейке выше на странице <a href="https://drive.google.com/">https://drive.google.com/</a> появится новый файл под названием Sample file.

### Скачивание данных из файла на Диске в Python

Скачайте файл, который мы загрузили выше.

```python
file_id = created.get('id') # Получаем ID ранее загруженного файла из объекта created, который был получен после загрузки

import io
from googleapiclient.http import MediaIoBaseDownload # Импортируем необходимые модули для работы с потоками и API Google

request = drive_service.files().get_media(fileId=file_id) # Создаём запрос для получения содержимого файла по его ID
downloaded = io.BytesIO() # Создаём объект BytesIO для временного хранения загружаемых данных в памяти
downloader = MediaIoBaseDownload(downloaded, request) # Создаём объект MediaIoBaseDownload для обработки загрузки файла
done = False # Переменная, чтобы отслеживать статус загрузки
while done is False:
  # Входим в цикл, пока файл не будет полностью загружен
  # _ — это заполнитель для объекта прогресса, который мы игнорируем.
    # (Наш файл небольшой, поэтому мы пропускаем отчет о прогрессе.)
  _, done = downloader.next_chunk() # Загружаем следующую часть файла

downloaded.seek(0) # Устанавливаем курсор в начале потока, чтобы прочитать загруженные данные
print('Downloaded file contents are: {}'.format(downloaded.read())) # Выводим содержимое загруженного файла на экран
```

Чтобы скачать другой файл, приведите переменную <code>file&#95;id</code> в соответствие с его идентификатором, таким как: 1uBtlaggVyWshwcyP6kEI-y&#95;W3P8D26sz.
